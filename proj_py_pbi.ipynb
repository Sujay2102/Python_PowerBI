```python
import pandas as pd
df=pd.read_csv("C:\\Users\\patil\\Downloads\\SalesProject.csv",encoding="latin-1")
#encoding-format of file-default UTF-8
df.head()
#to display first 5 records
```

```python
#check for null values
print(df.isnull().sum())
#isnull()-cheks for null values in each column 
#sum()-gets the total of null values,column wise
```

```python
print(df.info())
#To print information and check datatype
```

```python
#change datatype to datetime
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
print(df.info())
```

```python
#remove outliers-filter out data
#outliers-too small or too large values to removed.
df=df[(df['Sales']>25) & (df['Sales']<100000)]
#&-logical operator and 
print(df.info())
```

```python
import datetime as dt
df['Order Year']=df['Order Date'].dt.year#year will get the year part of date
df['Order Month']=df['Order Date'].dt.month#month will get the month number
print(df.info())
```

```python
#transform to multiple dataframes
customers=df[['Customer ID','Customer Name','Segment','Country','City','State','Postal Code','Region']]
#remove duplicates
customers=customers.drop_duplicates(subset='Customer ID')
print(customers.info())
```

```python
products=df[['Product ID','Category','Sub-Category','Product Name']]
products=products.drop_duplicates(subset='Product ID')
print(products.info())
```

```python
orders=df[['Row ID','Order ID','Order Date','Ship Date','Ship Mode','Customer ID','Product ID','Sales','Quantity','Discount','Profit','Order Year','Order Month']]
print(orders.info())
#no need to remove duplicates.
```
